```python
colour={"black":0,"brown":1,"red":2,"orange":3,"yellow":4,"violet":5,"green":6,"gray":7,"blue":8,"white":9}
power={"black":1,"brown":10,"red":100,"orange":1000,"yellow":10000,"violet":100000,"green":1000000,"gray":10000000,"blue":100000000,"white":1000000000}
def resistor(col1,col2,col3):
  d1=colour[col1]
  d2=colour[col2]
  d3=power[col3]
  c=(d1*10+d2)*d3
  return c
a=input("enter the 1st colour:")
b=input("enter the 2nd colour:")
d=input("enter the 3 rd colour:")
print("the resistance value is :",resistor(a,b,d))
```
